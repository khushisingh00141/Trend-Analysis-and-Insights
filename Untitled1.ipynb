```python

# WEEK 3: TREND ANALYSIS AND INSIGHTS GENERATION

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load cleaned dataset
df = pd.read_csv("Marketing_Campaign_Cleaned.csv")

# Convert dates
df["Start_Date"] = pd.to_datetime(
    df["Start_Date"], dayfirst=True, errors="coerce"
)

df["End_Date"] = pd.to_datetime(
    df["End_Date"], dayfirst=True, errors="coerce"
)

# Display basic information
print("Dataset Shape:", df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nChannel Categories:")
print(df["Channel"].value_counts())

print("\nStatistical Summary:")
display(df.describe().round(2))
```

**Monthly campaign-start **


```python

# Extract month from campaign start date
df["Start_Month"] = df["Start_Date"].dt.to_period("M")

# Monthly campaign count
monthly_campaigns = (
    df.dropna(subset=["Start_Month"])
      .groupby("Start_Month")
      .size()
      .reset_index(name="Campaign_Count")
)

monthly_campaigns["Start_Month"] = (
    monthly_campaigns["Start_Month"].astype(str)
)

display(monthly_campaigns)

# Visualize monthly campaign starts
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_campaigns["Start_Month"],
    monthly_campaigns["Campaign_Count"],
    marker="o"
)

plt.title("Monthly Campaign Start Trend - 2023")
plt.xlabel("Month")
plt.ylabel("Number of Campaigns")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# STEP 3: MONTHLY MARKETING PERFORMANCE ANALYSIS

# Make sure dates are in datetime format
df["Start_Date"] = pd.to_datetime(
    df["Start_Date"], dayfirst=True, errors="coerce"
)

# Create monthly grouping column
df["Month"] = df["Start_Date"].dt.to_period("M")

# Keep records with valid start dates
valid_df = df.dropna(subset=["Month"]).copy()

# Create a flag for records with known conversion values
valid_df["Conversion_Known"] = valid_df["Conversions"].notna()

# Monthly performance summary
monthly_performance = (
    valid_df.groupby("Month")
    .agg(
        Campaigns=("Campaign_ID", "count"),
        Impressions=("Impressions", "sum"),
        Clicks=("Clicks", "sum"),
        Spend=("Spend", "sum"),
        Conversions=("Conversions", "sum"),
        Known_Conversion_Records=("Conversion_Known", "sum"),
        Known_Conversion_Clicks=(
            "Clicks",
            lambda x: x[valid_df.loc[x.index, "Conversion_Known"]].sum()
        )
    )
    .reset_index()
)

# Calculate monthly CTR using totals
monthly_performance["CTR_%"] = (
    monthly_performance["Clicks"] /
    monthly_performance["Impressions"] * 100
)

# Calculate conversion rate only on records with known conversions
monthly_performance["Conversion_Rate_%"] = (
    monthly_performance["Conversions"] /
    monthly_performance["Known_Conversion_Clicks"] * 100
)

# Format month
monthly_performance["Month"] = (
    monthly_performance["Month"].astype(str)
)

# Display results
display(monthly_performance.round(2))
```

```python

# STEP 3.2: MONTHLY PERFORMANCE VISUALIZATION

metrics = [
    ("Impressions", "Monthly Impressions"),
    ("Clicks", "Monthly Clicks"),
    ("Spend", "Monthly Marketing Spend"),
    ("Conversions", "Monthly Recorded Conversions")
]

for column, chart_title in metrics:

    plt.figure(figsize=(12, 5))

    plt.plot(
        monthly_performance["Month"],
        monthly_performance[column],
        marker="o"
    )

    plt.title(chart_title)
    plt.xlabel("Month")
    plt.ylabel(column)
    plt.xticks(rotation=45)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
# STEP 4: CHANNEL-WISE PERFORMANCE ANALYSIS

# Create a flag for records with known conversion values
valid_df["Conversion_Known"] = valid_df["Conversions"].notna()

# Calculate channel-wise performance
channel_performance = (
    valid_df.groupby("Channel")
    .agg(
        Campaigns=("Campaign_ID", "count"),
        Impressions=("Impressions", "sum"),
        Clicks=("Clicks", "sum"),
        Spend=("Spend", "sum"),
        Conversions=("Conversions", "sum"),
        Known_Conversion_Records=("Conversion_Known", "sum"),
        Known_Conversion_Clicks=(
            "Clicks",
            lambda x: x[
                valid_df.loc[x.index, "Conversion_Known"]
            ].sum()
        )
    )
    .reset_index()
)

# Calculate weighted CTR
channel_performance["CTR_%"] = (
    channel_performance["Clicks"] /
    channel_performance["Impressions"] * 100
)

# Calculate conversion rate using records with known conversions
channel_performance["Conversion_Rate_%"] = (
    channel_performance["Conversions"] /
    channel_performance["Known_Conversion_Clicks"] * 100
)

# Display the result
display(channel_performance.round(2))
```

```python
# STEP 5: CORRELATION ANALYSIS

# Select numerical columns for correlation
correlation_data = valid_df[
    [
        "Impressions",
        "Clicks",
        "Spend",
        "Conversions",
        "Campaign_Duration_Days"
    ]
]

# Calculate Pearson correlation
correlation_matrix = correlation_data.corr(method="pearson")

# Display correlation values
display(correlation_matrix.round(2))
```

```python
# Create correlation heatmap

plt.figure(figsize=(9, 6))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    interpolation="nearest",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

# Add correlation values inside the heatmap
for i in range(len(correlation_matrix.columns)):
    for j in range(len(correlation_matrix.columns)):
        plt.text(
            j, i,
            round(correlation_matrix.iloc[i, j], 2),
            ha="center",
            va="center"
        )

plt.title("Correlation Heatmap of Marketing Campaign Metrics")
plt.tight_layout()
plt.show()
```

```python
# STEP 6: STATISTICAL ANALYSIS

numeric_columns = [
    "Impressions",
    "Clicks",
    "Spend",
    "Conversions",
    "Campaign_Duration_Days",
    "CTR_%",
    "Conversion_Rate_%"
]

# Generate descriptive statistics
statistics_summary = valid_df[numeric_columns].describe().T

statistics_summary = statistics_summary.round(2)

display(statistics_summary)
```

```python
# Detect outliers using IQR method

outlier_results = []

columns_to_check = [
    "Impressions",
    "Clicks",
    "Spend",
    "Conversions",
    "Campaign_Duration_Days"
]

for column in columns_to_check:

    data = valid_df[column].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = data[
        (data < lower_limit) |
        (data > upper_limit)
    ]

    outlier_results.append({
        "Metric": column,
        "Lower_Limit": lower_limit,
        "Upper_Limit": upper_limit,
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

```python
# Box plots for campaign metrics

for column in columns_to_check:

    plt.figure(figsize=(8, 4))

    plt.boxplot(valid_df[column].dropna())

    plt.title(f"Outlier Detection - {column}")
    plt.ylabel(column)

    plt.tight_layout()
    plt.show()
```
